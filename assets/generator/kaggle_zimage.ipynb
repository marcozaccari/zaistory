```python
# Percorsi e parametri. Il manifest viene cercato da solo tra i dataset
# collegati; se non ce n'e' nessuno si ripiega sul campione incorporato.
import os

# Deve stare PRIMA di importare torch: riduce la frammentazione della VRAM,
# che sulla T4 e' la differenza tra generare e andare in out of memory.
os.environ.setdefault('PYTORCH_CUDA_ALLOC_CONF', 'expandable_segments:True')

import glob, json, pathlib

OUT_DIR   = pathlib.Path('/kaggle/working/assets_out')
LIMIT     = None      # None = tutti i job del manifest; per le prove metti 6
MAX_SIDE  = 768       # tetto al lato lungo: a 1024 la T4 va in OOM
OFFLOAD   = 'model'   # 'model' = veloce; 'sequential' = pochissima VRAM, lento
OOM_FALLBACK_SIDE = 512   # riprova cosi' se un job va comunque in OOM

candidates = sorted(glob.glob('/kaggle/input/**/*manifest*.json', recursive=True))
if candidates:
    MANIFEST_PATH = candidates[0]
    manifest = json.loads(pathlib.Path(MANIFEST_PATH).read_text(encoding='utf-8'))
else:
    MANIFEST_PATH = '(campione incorporato nel notebook)'
    manifest = json.loads(EMBEDDED_SAMPLE)
jobs = manifest['jobs'][:LIMIT] if LIMIT else manifest['jobs']

for job in jobs:
    scale = min(1.0, MAX_SIDE / max(job['width'], job['height']))
    if scale < 1.0:
        # multipli di 64: i modelli a diffusione lavorano su latenti /8
        job['width']  = int(job['width'] * scale) // 64 * 64
        job['height'] = int(job['height'] * scale) // 64 * 64

print('manifest :', MANIFEST_PATH)
print('storia   :', manifest.get('title'))
print('job      :', len(jobs), 'su', manifest['counts']['total'])
print('misura   :', jobs[0]['width'], 'x', jobs[0]['height'])
print('stile    :', (manifest.get('style_suffix') or '')[:120])
```

## 3. Modello

**bf16 obbligatorio.** Provato sulla T4: in fp16 il transformer va in NaN
gia' a livello di latenti e le immagini escono nere. Sulla T4 il bf16 e'
emulato via software, quindi piu' lento, ma numericamente stabile.

Un solo caricamento per run: il modello occupa ~12 GB di RAM di sistema e
caricarne due in sequenza fa morire il kernel. Se lo smoke test fallisce si
cambiano `DTYPE`/`VAE_FP32` in cima alla cella e si rilancia.


```python
import numpy as np
import torch
from diffusers import ZImagePipeline

MODEL_ID = 'Tongyi-MAI/Z-Image-Turbo'

# Verificato sulla T4 di Kaggle il 29/08: in float16 il TRANSFORMER va in NaN
# (i latenti escono gia' non finiti, prima ancora del VAE), e le immagini
# arrivano tutte nere. Quindi bf16, che sulla T4 e' emulato via software e piu'
# lento, ma ha lo stesso intervallo di esponenti del fp32 e non satura.
#
# Un solo caricamento per run: il modello pesa ~12 GB in RAM e provarne due in
# sequenza fa morire il kernel per esaurimento memoria di sistema, non VRAM.
DTYPE = {'bf16': torch.bfloat16, 'fp16': torch.float16}['bf16']
VAE_FP32 = False    # utile solo se a saturare fosse la decodifica, non il caso qui

major, minor = torch.cuda.get_device_capability(0)
total_vram = torch.cuda.get_device_properties(0).total_memory / 2**30
print(f'GPU: {torch.cuda.get_device_name(0)} (sm_{major}{minor}, {total_vram:.1f} GB)')
print('dtype:', DTYPE, '(emulato via software)' if major < 8 and DTYPE is torch.bfloat16 else '')


def try_optim(obj, name, label):
    """Le ottimizzazioni di memoria non esistono su tutte le pipeline.
    Se mancano si tira dritto: non si fa fallire un run per una di queste."""
    fn = getattr(obj, name, None)
    if not callable(fn):
        print(f'  - {label}: non disponibile')
        return
    try:
        fn()
        print(f'  + {label}: attivo')
    except Exception as exc:
        print(f'  - {label}: fallito ({exc})')


pipe = ZImagePipeline.from_pretrained(MODEL_ID, dtype=DTYPE)
if VAE_FP32 and getattr(pipe, 'vae', None) is not None:
    pipe.vae.to(torch.float32)   # prima dell'offload: dopo, gli hook lo riportano indietro

if OFFLOAD == 'sequential':
    pipe.enable_sequential_cpu_offload()
else:
    pipe.enable_model_cpu_offload()

try_optim(pipe, 'enable_attention_slicing', 'attention slicing')
vae = getattr(pipe, 'vae', None)
if vae is not None:
    try_optim(vae, 'enable_slicing', 'vae slicing')
    try_optim(vae, 'enable_tiling', 'vae tiling')
pipe.set_progress_bar_config(disable=True)

# Smoke test: 512x512, prima i latenti e poi l'immagine, per sapere in trenta
# secondi se questa configurazione produce NaN invece di scoprirlo dopo 87
# immagini nere. Se fallisce NON si ricarica un'altra pipeline qui dentro:
# si cambia DTYPE/VAE_FP32 qui sopra e si rilancia il run.
kw = dict(prompt='a stone on a wooden table', width=512, height=512,
          num_inference_steps=8, guidance_scale=0.0,
          generator=torch.Generator('cuda').manual_seed(0))
latent_ok = bool(torch.isfinite(pipe(output_type='latent', **kw).images).all())
probe = pipe(output_type='np', **kw).images[0]
image_ok = bool(np.isfinite(probe).all()) and float(probe.std()) > 0.01
print(f'smoke test -> latenti finiti: {latent_ok} | immagine valida: {image_ok}')

if not (latent_ok and image_ok):
    colpa = 'il transformer satura' if not latent_ok else 'satura la decodifica VAE'
    rimedio = "prova DTYPE='fp16'" if not latent_ok else 'metti VAE_FP32 = True'
    raise RuntimeError(f'Immagini non valide: {colpa}. Rimedio: {rimedio}, '
                       'poi rilancia il run (non ricaricare qui: la RAM non basta per due modelli).')

print('modello pronto')
```

## 4. Generazione

Ogni job ha il suo `seed`, quindi il risultato e' riproducibile: se un'immagine
non convince, si cambia il seed di quel job nel manifest e si rilancia solo lui.
Il ciclo salta i file gia' presenti, cosi' una sessione interrotta si riprende.


```python
import gc, time, traceback
import numpy as np
from PIL import Image

OUT_DIR.mkdir(parents=True, exist_ok=True)
report, t0 = [], time.time()
MAX_INVALID_IN_A_ROW = 2   # meglio fermarsi che riempire lo zip di immagini nere


def free_vram():
    """Senza questa, il primo OOM lascia la VRAM sporca e fa cadere a valanga
    tutti i job successivi."""
    gc.collect()
    torch.cuda.empty_cache()


def render(job, width, height):
    """Ritorna (immagine_PIL, valida). Si lavora in numpy per intercettare i
    NaN prima della conversione a uint8, che li trasformerebbe in nero."""
    arr = pipe(
        job['prompt'],
        width=width, height=height,
        num_inference_steps=job['steps'],
        guidance_scale=job['cfg'],
        generator=torch.Generator('cuda').manual_seed(job['seed']),
        output_type='np',
    ).images[0]
    valid = bool(np.isfinite(arr).all()) and float(arr.std()) > 0.01
    image = Image.fromarray((np.nan_to_num(arr) * 255).round().astype('uint8'))
    return image, valid


invalid_streak = 0

for i, job in enumerate(jobs, 1):
    dest = OUT_DIR / job['file']
    if dest.exists():
        print(f"[{i}/{len(jobs)}] {job['id']} - gia' fatto, salto")
        report.append({'id': job['id'], 'status': 'skipped'})
        continue
    dest.parent.mkdir(parents=True, exist_ok=True)

    started = time.time()
    width, height = job['width'], job['height']
    image, valid, err = None, False, None

    for attempt in (1, 2):
        try:
            image, valid = render(job, width, height)
            break
        except torch.OutOfMemoryError as exc:
            err = exc
            free_vram()
            if attempt == 1 and OOM_FALLBACK_SIDE < max(width, height):
                scale = OOM_FALLBACK_SIDE / max(width, height)
                width, height = int(width * scale) // 64 * 64, int(height * scale) // 64 * 64
                print(f"    OOM: riprovo {job['id']} a {width}x{height}")
            else:
                break
        except Exception as exc:
            err = exc
            traceback.print_exc()
            break

    if image is None:
        report.append({'id': job['id'], 'status': 'error', 'error': repr(err)})
        print(f"[{i}/{len(jobs)}] {job['id']} - ERRORE: {err}")
        free_vram()
        continue

    dt = time.time() - started
    if not valid:
        invalid_streak += 1
        report.append({'id': job['id'], 'status': 'invalid', 'seconds': round(dt, 1)})
        print(f"[{i}/{len(jobs)}] {job['id']} - NON VALIDA (nera/NaN), non salvata")
        free_vram()
        if invalid_streak >= MAX_INVALID_IN_A_ROW:
            print(f"\nInterrotto: {invalid_streak} immagini non valide di fila. "
                  "Non e' un problema di prompt, e' la configurazione numerica: "
                  "rilancia la cella del modello e guarda quale dtype passa lo smoke test.")
            break
        continue

    invalid_streak = 0
    image.save(dest)
    report.append({'id': job['id'], 'status': 'ok', 'seconds': round(dt, 1),
                   'file': job['file'], 'size': f'{width}x{height}'})
    print(f"[{i}/{len(jobs)}] {job['id']} - {dt:.1f}s  {width}x{height}")
    free_vram()

ok = sum(1 for r in report if r['status'] == 'ok')
print(f"\nfatte {ok}/{len(jobs)} in {(time.time()-t0)/60:.1f} minuti")
(OUT_DIR / 'generation_report.json').write_text(json.dumps(report, ensure_ascii=False, indent=2))
```

## 5. Provino a contatto

Serve a giudicare le immagini senza scaricare lo zip: qui si decide quali ancore
tenere e quali rifare cambiando seed.


```python
import matplotlib.pyplot as plt
from PIL import Image

done = [r for r in report if r['status'] == 'ok']
if done:
    cols = min(3, len(done))
    rows = (len(done) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 5.6 * rows))
    for ax, r in zip(np.atleast_1d(axes).ravel(), done):
        ax.imshow(Image.open(OUT_DIR / r['file']))
        ax.set_title(r['id'], fontsize=8)
        ax.axis('off')
    for ax in np.atleast_1d(axes).ravel()[len(done):]:
        ax.axis('off')
    plt.tight_layout()
    plt.show()
else:
    print('nessuna immagine da mostrare')
```
